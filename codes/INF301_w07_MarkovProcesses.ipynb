# Probabilistic Systems

Time is traditionally partitioned into three sets: past, present, and future.

We are aware of what is happening at this moment, and fortunately, we remember the past. The idea is to predict the future.

To achieve this goal, a discrete time $k$ is considered to represent the present, $k–1$, $k–2$, ... to represent the past, and $k+1$ to describe the immediate future.

<center>

| | &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Past &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;  | Present  |&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; &nbsp;&nbsp;&nbsp; Future &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;  |
|:-------|:-------------------------:|:--------:|:------------------------:|
| Time   | $k-1, k-2, \dots$         | $k$      | $k+1, k+2, \dots$        |
| Input  | $x(k-1), x(k-2), \dots$   | $x(k)$   | $x(k+1), x(k+2), \dots$  |
| State  | $y(k-1), y(k-2), \dots$   | $y(k)$   | $y(k+1), y(k+2), \dots$  |
| Output | $z(k-1), z(k-2), \dots$   | $z(k)$   | $z(k+1), z(k+2), \dots$  |
</center>


Generally, the future state depends on the history of the input signals and the sequence of states.

If the future is deterministic, it is predictable, given the previous history and $y(k+1) = f[x(0), x(1), x(2), ...; y(0), y(1), y(2), ...]$ for a given function $f$.
The problem is that the future is often non-deterministic, necessitating probabilistic assumptions.

Thus, a probabilistic model makes a conjecture about the future state based on the present and past history of states and input signals.

For inputs $x(k)$ and states $y(k)$, the most general probabilistic model is $Pr[f(y(k+1) | g(x(k), x(k-1), x(k-2), ...), h(y(k), y(k-1), y(k-2), ...)]$, where $f$, $g$ and $h$ are carefully defined functions.

Here, the simpler models are considered, and the following can be assumed:
* P1: The system has a finite number of states $j = 1, 2, ..., n$.
* P2: The future state depends only on the current state.
* P3: If $p_j(k)$ is the probability that the system is in state $j$, we can write:
$$ p_{j}(k+1)=p_{1 j} p_{1}(k)+p_{2 j} p_{2}(k)+\cdots+p_{n j} p_{n}(k) $$

# Markov Systems

The systems described above are known as Markov systems, and the probabilistic models that characterize them are referred to as Markov Chains.

Since Markov chains are finite, it is possible to characterize the state of the chain using a vector.

Thus, assuming $n$ states, $p(k)$ has $n$ elements, with the $j$-th component of this vector being the probability $p_j(k)$ that the system is in state $j$ at time $k$.

$$ \mathbf{p}(k)=\left[p_{1}(k), p_{2}(k), \cdots, p_{n}(k)\right] $$

Because of linearity, the set of transition probabilities $p_{ij}$ from the current state to any other is modeled by the transition matrix.

$$ \mathbf{P}=\left[\begin{array}{cccc} p_{11} & p_{12} & \cdots & p_{1 n} \\ p_{21} & p_{21} & \cdots & p_{2 n} \\ \vdots & \vdots & \ddots & \vdots \\ p_{n 1} & p_{n 2} & \cdots & p_{n n} \end{array}\right] $$

In this way, one can describe the future state in a Markov chain using vector and matrix notation:

$$ p(k+1)=p(k)\mathbf{P} $$

# Discrete Markov Processes

Since a Markov process has $n$ finite states, a convenient graphical representation of this process is a state diagram.

Each of the $n$ states is represented by a vertical line, and directed arcs connecting the vertical lines represent the transition probabilities.

# Steady State

Since many problems deal only with steady-state behavior, one must obtain these solutions. By definition, the solution to the probability transition matrix and the state vector is:

$$ \begin{aligned} &\mathbf{P}_{s s}=\lim _{k \rightarrow \infty} \mathbf{P}^{k}=\left[\begin{array}{lll} 1 & 0 & 0 \\ 1 & 0 & 0 \\ 1 & 0 & 0 \end{array}\right] \\ &\pi=\lim _{k \rightarrow \infty} \mathbf{p}(k)=\left[\begin{array}{lll} 1 & 0 & 0 \end{array}\right] \end{aligned} $$

One cannot always find the desired solution by inspection; however, since it is a first-order initial value system, it can be calculated with relative ease.

````
import numpy as np

T = np.zeros(4); T[0] = 1
y = 1; n = 10000

for i in range(n):
    if y == 0:
        r = np.floor(2*np.random.random_sample())
        if r == 0:
           yf = 0
        else:
           yf = 1
    elif y == 1:
        r = np.floor(4*np.random.random_sample())
        if r == 0:
          yf = 0
        elif r == 2:
          yf = 2
        else:
          yf = 1
    elif y == 2:
        r = np.floor(4*np.random.random_sample())
        if r == 1:
          yf = 1
        elif r == 3:
          yf = 3
        else:
          yf = 2
    elif y == 3:
        r = np.floor(2*np.random.random_sample())
        if r == 0:
          yf = 2
        else:
          yf = 3
    T[yf] += 1;
    y = yf

Pi = T/n
print("Pi = ",Pi)
````

# Poisson Processes

So far, only discrete-time Markov chains have been considered, but it is also possible to use them for event-driven (asynchronous) models in continuous time.

Considering the example of the telephone system, each telephone call is an event that occurs randomly in continuous time, and the exact time of occurrence is unknown.

However, the time elapsed between events can be determined statistically, and if it follows an exponential distribution, the process is said to be Poissonian.

Suppose $N(t)$ is a random variable that represents the number of events $n$ occurring in the continuous time interval $[0, t]$.

We define $P_n(t) = P_r[N(t) = n]$ as the probability that exactly $n$ events occur in the interval $[0, t]$. Three postulates can characterize this probability:

* The number of events occurring in a given interval is independent of the number of events in any other interval.
* The probability that a single event will occur in a short time interval $[t, t+h]$ is approximately proportional to $h$.
* The probability that more than one event will occur in a short time interval $[t, t+h]$ is zero.

Any random variable that satisfies these three postulates is called a Poisson random variable, and $P_n(t)$ is called a Poisson process.
From careful analysis of the postulates, we obtain:

$$ P_{0}(t)=e^{-\lambda t} $$

Extending to $n$ events, the general solution is:
$$ P_{n}(t)=\frac{(\lambda t)^{n}}{n !} e^{-\lambda t} $$

It is sometimes convenient to replace $\lambda t$ with $\mu$. While $\lambda$ represents the average rate of occurrences, $\mu$ is the average number of events during time $t$.


# Properties of Poisson Processes

Although explicit probabilities given by $P_n(t)$ are useful, data such as mean and variance are often sufficient.

Interestingly, both the mean and variance of a Poisson distribution are identical and equal to $\mu = \sigma^2 = \lambda t$.

This property is unique to Poisson distributions.

## Example 5

Consider the specific Poisson distribution characterized by $\lambda = 0.9$ events per time unit. Graph and interpret the Poisson probability values and the distribution functions.

**Solution**

The Poisson probability function $P_n(t)$ depends on two variables: $n$ and $t$. Therefore, a three-dimensional approach is required to sketch the graphs, which are shown for $t \in [1, \dots, 5]$ and $n \in [1, \dots, 9]$

````
import numpy as np
import math
import matplotlib.pyplot as plt

# Parameters
lambda_ = 0.9
t_values = np.arange(1, 6)     # t = 1, 2, 3, 4, 5
n_values = np.arange(0, 10)    # n = 0, 1, ..., 9

# 3D Figure
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

# Generation of Pn(t) curves for different values of t
for t in t_values:
    P = (lambda_ * t)**n_values / np.vectorize(math.factorial)(n_values) * np.exp(-lambda_ * t)
    # Constant t-axis (line along n)
    t_line = np.full_like(n_values, t)
    ax.plot(t_line, n_values, P, lw=4, alpha=0.8, label=f't={t}')

# Labels and adjustments
ax.set_xticks(t_values)
ax.set_xlabel(r'$t$')
ax.set_xlim([1, 5])
ax.set_ylabel(r'$n$')
ax.set_ylim([0, 9])
ax.set_zlabel(r'$P_n(t)$',labelpad=1)
ax.view_init(elev=25, azim=35)
ax.legend(bbox_to_anchor=(0.7, 0.75), loc='upper left', borderaxespad=0.)
plt.tight_layout()
plt.show()
````

# Exponential Distribution

The Poisson distribution is defined for a discrete random variable $N(t)$ and is related to another (but continuous) random variable called the exponential random variable.

Instead of looking at the number of events within a fixed time interval, the exponential distribution considers the time $T$ between each event, known as the inter-event time.

Since it is a Poisson process and no events occur in the interval $[0, t]$:

$$ \operatorname{Pr}[T>t]=\operatorname{Pr}[N=0]=P_{0}(t)=e^{-\lambda t} $$

The relationship between the Poisson and exponential distributions is especially important because they are dual to each other.

The Poisson distribution measures the number of events that occur in the interval $[0, t]$, and the exponential distribution measures the time between these events.

In statistical terms, the mean of an exponential distribution is given by $1/λ$, while the variance is given by $1/λ^2$. The statistics are summarized in the following table:

<center>

|        |         Poisson distribution           |    Exponential distribution     |
| :----- |:----------------------:| :--------------: |
| Measurement   | Number of events $n$ durint time $t$       | Time $t$ between each event                |
| Type  | Discrete  | Continuous             |
| Mean  | $\mu = \lambda t$  | $\mu = 1/\lambda$              |
| Variance | $\sigma^2 = \lambda t$  | $\sigma^2 = 1/\lambda^2$              |
</center>

Thus, the times of a distribution like this can be obtained using the formula $t^{\mathrm{new}} = t – μ\times \ln(\mathrm{RND})$, where $μ = 1/λ$ and $\mathrm{RND}$ is a random number uniformly distributed in the interval $[0, 1[$.

Use simulation to obtain the results of Example 6, which have already been obtained analytically.

The results to be obtained are the average number of calls in a 5-minute interval and the probabilities of having 0 calls and more than 1 call in a 1-minute interval, considering $\lambda$ = 2 calls/minute.

To this end, a 24-hour interval (1440 minutes) is considered for statistical analysis. It is necessary to compute calls in 1 and 5-minute intervals, and the simulation script used is presented next.



````
import numpy as np

def container(hTime, tTime, d):
  it = np.zeros(int(hTime/d)+1)
  it[1:] = d
  it = np.cumsum(it)

  c = np.zeros(len(it) - 1)
  j = 1;  ct = tTime[1]
  for i in range(len(c)):
    while (ct <= it[i+1]):
        if (tTime[j] <= it[i+1]):
          j += 1
          ct = tTime[j]
          c[i] += 1
  return c

# simulation of a 24-hour day, equal to 1440 minutes
# For a more refined simulation, an average of several days should be taken

_lambda = 2; _mu = 1/_lambda; _time = 1440
t = np.array([0])
while (t[-1] < _time):
  t1 = t[-1] -_mu*np.log(1.0-np.random.random_sample())
  t = np.hstack((t,t1))

# calculation of the number of calls per interval
interval5 = 5
c5 = container(_time,t,interval5)
interval1 = 1
c1 = container(_time,t,interval1)

# statistics
a5 = np.mean(c5)
a1 = np.mean(c1)

n0 = np.where(c1 == 0); p0 = len(n0[0])/len(c1)
n1 = np.where(c1 == 1); p1 = len(n1[0])/len(c1)
nn = np.where(c1 > 1); pn = len(nn[0])/len(c1)

print("a) An average of %.2f calls is expected in %.1f minutes."%(a5,interval5))
print("b) The probability of having no calls is %.4f."%p0)
print("c) The probability that more than one call will be received in the interval 14:03 to 14:04 is %.4f."%pn)
````

## Birth and Death Process

There is a special case of Markov chains where the transitions occur only between a given state and its adjacent states. Such systems are called birth-and-death systems.

The reason is that, considering the state as the number of individuals in a population, the state rises or falls as individuals are born (increment) or die (decrement).

For any state $k$:
* Birth process: the arc goes from the current state to a higher one;
* Death process: the arc goes from the current state to a lower one;
* Retention process: the arc goes from the current state to itself.

The retention arcs can be inferred, since the sum of all arcs leaving a given state is zero. By ignoring the retention arcs, a simplified intensity diagram can be obtained, which is especially useful for steady-state problems because, at each node, the sum of the inputs equals the sum of the outputs.

From the simplified diagram, we can write the balance equation at node $k$:

$$\lambda_{k-1}\pi_{k-1}+ \mu_{k+1}\pi_{k+1}=(\lambda_k+\mu_k)\pi_k,$$

for birth rates $\lambda_i$, deaths $\mu_i$ and steady-state transition probabilities $\pi_i$.

````
import numpy as np

Pit = np.array([9/37, 12/37, 16/37])

T = np.zeros(3); T[0] = 1
y = 1; n = 100000

for i in range(n):
    if y == 0:
        r = np.random.random_sample()
        if r < 1/3:
           yf = 1
        else:
           yf = 0
    elif y == 1:
        r = np.random.random_sample()
        if r < 1/4:
          yf = 0
        elif r < 1/4 + 1/3:
          yf = 2
        else:
          yf = 1
    elif y == 2:
        r = np.random.random_sample()
        if r < 1/4:
          yf = 1
        else:
          yf = 2
    T[yf] += 1;
    y = yf

Pi = T/n
print("Pi (simulation) = ",Pi)
print("Pi (theory) = ",Pit)
````
